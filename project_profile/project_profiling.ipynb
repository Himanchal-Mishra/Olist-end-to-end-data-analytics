```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load datasets
customers = pd.read_csv("olist_customers_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
category_translation = pd.read_csv("product_category_name_translation.csv")
```

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers,
    "geolocation": geolocation,
    "category_translation": category_translation
}

summary = []

for name, df in datasets.items():
    summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Duplicate Rows": df.duplicated().sum(),
        "Total Missing Values": df.isnull().sum().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
def profile_dataset(df, name):
    print("=" * 60)
    print(f"DATASET: {name}")
    print("=" * 60)

    print(f"\nRows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nData Types:")
    print(df.dtypes)

    print("\nSample Records:")
    print(df.head())

    print("\n")
```

### Customers Table - Key Observations

* The customers dataset contains 99,441 records and 5 columns.
* No missing values or duplicate rows were identified.
* customer_id appears suitable as the primary business key.
* Customer location information is available at city and state level, enabling future geographic analysis.
* Overall, the dataset is clean and requires no preprocessing.


### Orders Table - Key Observations

* The orders dataset contains 99,441 records and captures the complete order lifecycle.
* Missing values exist in approval and delivery-related timestamps.
* These missing timestamps likely correspond to cancelled, unavailable, or undelivered orders rather than data quality issues.
* No duplicate records were found.
* The table will be critical for delivery performance and order fulfillment analysis.


### Remaining Tables - Key Observations

**Order Items**

* Contains 112,650 records with no missing values or duplicates.
* Represents product-level transactions and is a strong candidate for the fact table grain.

**Payments**

* Contains 103,886 payment records with complete data quality.
* Supports payment behavior and installment analysis.

**Reviews**

* Review scores are complete, but review titles and comments contain substantial missing values.
* Missing comments are expected because textual feedback is optional.
* review_id should not automatically be assumed to be unique and requires validation.

**Products**

* Most product records are complete.
* 610 products are missing category-related metadata and 2 products have missing dimensions.
* Missing values affect only a small proportion of records.

**Sellers**

* Clean dataset with no missing values or duplicates.
* Suitable for seller-level performance analysis.

**Geolocation**

* Contains over 1 million records.
* Significant duplicate records (261,831) were detected.
* May require deduplication before use and is lower priority for initial modeling.

**Category Translation**

* Small lookup table with complete data quality.
* Can be merged with product information during transformation.


```python


mql = pd.read_csv("olist_marketing_qualified_leads_dataset.csv")
closed_deals = pd.read_csv("olist_closed_deals_dataset.csv")

marketing_datasets = {
    "marketing_qualified_leads": mql,
    "closed_deals": closed_deals
}

for name, df in marketing_datasets.items():
    
    print("="*60)
    print(f"DATASET: {name}")
    print("="*60)

    print(f"\nRows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nData Types:")
    print(df.dtypes)

    print("\nSample Records:")
    print(df.head())

    print("\n")
```

### Marketing Qualified Leads (MQL) - Key Observations

* The dataset contains 8,000 marketing-qualified leads and 4 columns.
* mql_id appears to uniquely identify each lead.
* No duplicate records were found.
* Only 60 records have missing values in the origin column, representing less than 1% of the dataset.
* The dataset captures lead acquisition information, including first contact date, landing page, and acquisition source.
* Overall data quality is high and suitable for marketing funnel analysis.


### Closed Deals - Key Observations

* The dataset contains 842 successfully converted leads and 14 columns.
* mql_id can be used to link converted leads back to the marketing-qualified leads dataset.
* No duplicate records were detected.
* Several seller profiling fields contain substantial missing values, including company information, GTIN availability, stock levels, and product catalog size.
* Business segment, lead type, and business type are largely populated and remain useful for segmentation analysis.
* Despite missing seller attributes, the dataset is suitable for conversion, sales funnel, and lead quality analysis.


```python
print("MQL Unique IDs:", mql['mql_id'].nunique())
print("Closed Deal Unique IDs:", closed_deals['mql_id'].nunique())

print(
    "Conversion Rate:",
    round(
        closed_deals['mql_id'].nunique() /
        mql['mql_id'].nunique() * 100,
        2
    ),
    "%"
)
```

### Marketing Funnel Insights

* The Marketing Qualified Leads (MQL) dataset contains 8,000 leads generated through various acquisition channels.
* The Closed Deals dataset contains 842 successfully converted leads.
* Based on the available data, the lead-to-deal conversion rate is approximately 10.53%.
* This indicates that roughly 1 out of every 10 qualified leads ultimately became a seller on the platform.
* The datasets can be linked using mql_id, enabling future analysis of lead sources, sales performance, and conversion effectiveness.
